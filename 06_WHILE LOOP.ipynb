```python
# Print the sum of all the numbers between 1 to 10 (both inclusive)
# 1. for loop
# 2. while loop
sum = 0
for i in range(1,11):
  sum+=i
  print(sum)
```

```python
# Take a number from the user and print their multiplication till its 10 multiples using while loop.
number = int(input("Enter your number:"))
i = 1
while i<=10:
  print(number,"*",i,"=",number*i)
  i+=1
```
