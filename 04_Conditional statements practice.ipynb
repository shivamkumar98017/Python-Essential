```python
# Body Mass Index (BMI) is a simple measure that helps determine whether a
# person's weight is appropriate for their height.
# Write a Python program that takes the BMI value as input from the user
# and displays the corresponding weight category using the following rules:
# BMI Value   Category
# Less than 18.5  Underweight
# 18.5 to 24.9
# Normal Weight
# 25.0 to 29.9
# Overweight
# 30.0 or above   Obese

BMI = float(input("Enter your BMI:"))
if BMI < 18.5 :
  print("Underweight")
elif BMI < 24.9:
  print("Normal Weight")
elif BMI <29.9:
  print("Over Weight")
else:
  print("Obses")
```

```python
# Write a Python program that takes an integer as input and checks its divisibility.
# Display the appropriate message based on the following conditions:
# If the number is divisible by both 3 and 5, print:
# Divisible by both 3 and 5
# If the number is divisible only by 3, print:
# Divisible only by 3
# If the number is divisible only by 5, print:
# Divisible only by 5
# Otherwise, print:
# Not divisible by 3 or 5

Number = int(input("Enter a number:"))
if Number%3==0:
  print("Divisible only by 3.")
elif Number%5==0:
  print("Divisible only by 5.")
elif Number%3==0 and Number%5==0:
  print("Divisible by both 3 and 5.")
else:
  print("Not divisible by 3 or 5.")
```
